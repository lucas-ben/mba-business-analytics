### **ETL processes**

```python
def load(data_frame, target_table):
    # loading data from sql
    data_frame.to_sql(name=target_table, con=POSTGRES_CONNECTION)
    print(f"Loading data to the {target_table} table")

def extract(file_name):
    print(f"Extracting data from {file_name}")
    return pd.read_csv(file_name)

extracted_data = extract(file_name="data.csv") # extracting the data from the raw data file
transformed_data = transform(data_frame=extracted_data) # transforming the extracted data
load(data_frame=transformed_data, target_table="cleaned_data") # loadingg the transformed data to a sql table
```
