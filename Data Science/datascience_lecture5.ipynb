Word vectors help us extract information from sentences (unstructured data) so I don't have to collect features myself. To create word vectors, I need some knowledge (need to understand the text I want to embed) and encoding mechanism to transform text into numbers.

Vectors are embeddings that represent unstructured data in a numerical representation of the words that show up in that structure, and we can use the numerical representation as a feature to feed into the model. 

With vectors, we can extract information from sentences so I don't have to manually structure the data myself.

GloVe is a good corpora available 

To start the process of word vectors, tokenize the text (this generally only works in English). There are tokenizer APIs available to tokenize text data in other languages

To begin, create embeddings by creating pairs of words with a **center** word and **context words**

We can use a window size to tell an embedding algorithm how many neighbouring words to look at for context. (ie: A window size of 1 mean that a center word only looks at its immediate neighbours)

When pairs are created, all possible pairs of a sentence are identified. Each word has its turn becoming a center word and context word.

Once there are word pairs, I need to convert these pairs to some numerical representation and this is done using one-hot encoding. 

The embeddings become inputs into my model.

Can create machine readable stories from structured data by leveraging LLMs to take the data and create a descriptive text. Then I can create word embeddings and proceed with these numeric inputs for my model.

```python
import nltk
# downloading corpora
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = "I really enjoyed the recipe this week. I especially liked the chicken, but the broccoli could have been better. I thought the butter was just right, but the maple syrup was too much."
```

```python
# tokenizing
sentences = nltk.sent_tokenize(text)
```

```python
sentences # now just a list separated by a comma
```

```python
words = nltk.word_tokenize(text)
```

```python
words
```

```python
# can use python functions to clean up text
words = [w.lower() for w in words if w.isalpha()] # makes every letter the same case
words
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
filtered_tokens = [word.lower() for word in words if word not in stop_words]
filtered_tokens # only keeps meaningful words 
```

```python
stemmer = nltk.PorterStemmer()
lemmatizer = nltk.WordNetLemmatizer()
```

```python
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
stemmed_tokens
```

```python
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
lemmatized_tokens
```
