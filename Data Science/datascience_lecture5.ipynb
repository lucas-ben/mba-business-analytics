## **NLP**

- Most common use case is **topic modelling**

    - Used by customer service teams

    - If we can go through our tickets and understand the context (the number of times certain tickets come up? how many calls does it take to resolve a customer problem?) we can solve problems

    - Understanding context to make decisions (directly related to cost management)

    - Foundational models are vector based representations of everything a company knows about a customer. When data is stored in vectors it is much easier to work with (we can use embeddings to represent a customer's profile). This makes it easier to insert into reinforcement learning methods. 

- Tools like ChatGPT use a predictive text system. But the computation capacity is huge.



NLP is best used for automation. Sales process at wealthsimple, say a major promotion then a sales person will give a customer a call within 7 days if they're within a certain level. Humans are limited. So they adopted RicoAI is an AI bot that calls people. The model is trained on promotions, products, and how to answer questions. Any sort of contact with a sales lead will increase the likelihood of a client depositing money, but they deposit less if it is not a human.

Automation that creates customer experiences at scale is what companies are doing by adopting AI. NLP is huge in this automation process.

Use ChatGPT to create some embeddings for data preprocessing for NLP



Challenges in NLP:

- Context is not always clear

- Words that sound the same (only relevant for audio NLP not text applications)

- Slang and colloquialisms

- Lack of useable data (LLMs just pull the data and are not vetting it to determine if it's correct for the situation)

Main algorithms today

- BERT by Google (Uses attention to understand context)

- XLNet by Google (understands longer dependencies than BERT)

- T-NLG by Microsoft (language generation)
