## **Convolutional Neural Networks**

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import to_categorical
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_train, X_test = X_train / 255.0, X_test / 255.0
```

```python
y_train, y_test = to_categorical(y_train), to_categorical(y_test)
```

```python
from tensorflow.keras import layers
```

```python
def simple_model(initializer, activation='relu'):
  return tf.keras.Sequential([
     Flatten(input_shape=(28, 28)),
     Dense(32, activation, kernel_initializer=initializer),
     Dense(32, activation, kernel_initializer=initializer),
     Dense(32, activation, kernel_initializer=initializer),
     Dense(10, activation='softmax', kernel_initializer=tf.keras.initializers.glorot_normal())
  ])
```

```python
init = tf.initializers.zeros()
activation = 'relu'
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
model_zeros = simple_model(init, activation)
model_zeros.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model_zeros.fit(X_train, y_train, epochs=10, batch_size=3200, validation_data=(X_test, y_test))
```

```python
init = tf.initializers.glorot_uniform()
activation = 'relu'
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
model_zeros = simple_model(init, activation)
model_zeros.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model_zeros.fit(X_train, y_train, epochs=10, batch_size=3200, validation_data=(X_test, y_test))
```

```python
init = tf.initializers.he_uniform()
activation = 'relu'
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
model_zeros = simple_model(init, activation)
model_zeros.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model_zeros.fit(X_train, y_train, epochs=10, batch_size=3200, validation_data=(X_test, y_test))
```

```python
# normalization layers

def simple_model_bn(initializer='he_uniform', activation='relu'):
  return tf.keras.Sequential([
     Flatten(input_shape=(28, 28)),
     layers.BatchNormalization(trainable=True, scale=True, center=True),
     Dense(32, activation, kernel_initializer=initializer),
     layers.BatchNormalization(trainable=True, scale=True, center=True),
     Dense(32, activation, kernel_initializer=initializer),
     layers.BatchNormalization(trainable=True, scale=True, center=True),
     Dense(32, activation, kernel_initializer=initializer),
     layers.BatchNormalization(trainable=True, scale=True, center=True),
     layers.Dense(10, activation='softmax', kernel_initializer=tf.keras.initializers.glorot_normal())
  ])
```

```python
init = tf.initializers.random_uniform()
activation = 'relu'
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
model_bn = simple_model_bn(init, activation)
model_bn.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model_bn.fit(X_train, y_train, epochs=10, batch_size=3200, validation_data=(X_test, y_test))
```

```python
# batch normalization significantly helps a neural net improve its accuracy more quickly
```

```python
init = tf.initializers.he_uniform()
activation = 'relu'
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)
model_bn = simple_model_bn(init, activation)
model_bn.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model_bn.fit(X_train, y_train, epochs=10, batch_size=3200, validation_data=(X_test, y_test))
```

```python
init = tf.initializers.he_uniform()
activation = 'relu'
optimizer = tf.keras.optimizers.Adagrad(learning_rate=0.1)
model_bn = simple_model_bn(init, activation)
model_bn.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])
model_bn.fit(X_train, y_train, epochs=10, batch_size=3200, validation_data=(X_test, y_test))
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# In the above we see that lower learning rate addressed the overfitting problem, but now we don't have enough training epochs
# so we can use early stopping

init = tf.initializers.he_uniform()
activation = 'relu'
optimizer = tf.keras.optimizers.Adam(learning_rate=0.0001)
model_bn = simple_model_bn(init, activation)
model_bn.compile(optimizer=optimizer, loss='categorical_crossentropy', metrics=['accuracy'])

es = EarlyStopping(monitor='val_loss', mode='min', patience=10, restore_best_weights=True, verbose=1)

model_bn.fit(X_train, y_train, epochs=2000, callbacks =[es], batch_size=3200, validation_data=(X_test, y_test), verbose=1)

_, train_acc = model_bn.evaluate(X_train, y_train, verbose=0)
_, test_acc = model_bn.evaluate(X_test, y_test, verbose=0)
print('Train: %.3f, Test: %.3f' % (train_acc, test_acc))
```
