```python
no_missing_cols = df.columns[df.isna().sum() == 0]
len(no_missing_cols)
```

```python
matrix = df[no_missing_cols]
```

```python
num_matrix = matrix.select_dtypes(include=['number'])

svd_matrix = np.linalg.svd(num_matrix) # singular value decomposition
```

```python
V_transpose = svd_matrix[2] # getting the right singular vectors
V = V_transpose.T
V
```

```python
pca = PCA()

pca.fit(num_matrix)

explained_variance = pca.explained_variance_ratio_

pca.components_
```

```python
sum(explained_variance[:4]) # 4 principal components explain 90% of the variation in the data
```
